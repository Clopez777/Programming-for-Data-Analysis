<h3>Qué hace pop() y cómo difiere

El método pop() sirve para eliminar elementos de una lista, pero tiene la particularidad de que devuelve el valor eliminado, por lo que se puede guardar en una variable o utilizarlo después. Además, permite eliminar varios elementos de una lista utilizando un rango de posiciones.

<h3>Defina un diccionario</h3>

````python
personas = {
    'Ferney' : (34, 85, 1.72, True),
    'Alison' :(24, 65, 1.60, True),
    'Yenny' :(32, 70, 1.58, False),
    'Viviana' : (34, 60, 1.60, False),
    'Miguel' : (15, 90, 1.70, True),
    'Wilmar' : (13, 70, 1.76, False)
}

print(personas)
````

<h3>Implemente ciclos (for o while)</h3>

````python
Edades = []
Pesos = []
Estaturas = []
Estudia = []

for nombre, datos in personas.items():
    Edades.append(datos[0])
    Pesos.append(datos[1])
    Estaturas.append(datos[2])
    Estudia.append(datos[3])

print (f'Edades: {Edades}')
print (f'Pesos: {Pesos}')
print (f'Estaturas: {Estaturas}')
print (f'Estudia: {Estudia}')
````

<h3>Defina una función que reciba una lista</h3>

````python
def promedio(lista):
    return sum(lista) / len(lista)


print(f'Promedio de edades:, {promedio(Edades)}')
print(f'Promedio de pesos: {promedio(Pesos)}')
print(f'Promedio de estaturas: {promedio(Estaturas)}')
print(f'Promedio de personas que estudian: {promedio(Estudia)}')
````

<h3> Conjuntos (sets)</h3>

````python
edades_unicas = set(Edades)
edades_referencia = [25, 44 ,32 ,27 ,30, 34, 15, 13, 18]

print(edades_unicas.union(edades_referencia))
print(edades_unicas.intersection(edades_referencia))
print(edades_unicas.difference(edades_referencia))

````

<h3>Diccionario de clasificacion con claves Mayores de edad y
Menores de edad. </h3>

````python
clasificacion = {
    'Mayores de edad': [] ,
    'Menores de edad': []
}

for edad in Edades:
     if edad >=18:
         clasificacion['Mayores de edad'].append(edad)
     else:
         clasificacion['Menores de edad'].append(edad)

print(clasificacion)
````

El código permite clasificar las edades de una lista en dos grupos utilizando un diccionario, un ciclo for y una condición if. Las personas que tienen 18 años o más se guardan en la lista de mayores de edad, mientras que las personas menores de 18 años se guardan en la lista de menores de edad.

<h3>List/dict comprehensions</h3>

````python
Edades = [datos[0] for nombre, datos in personas.items()]
Pesos = [datos[1] for nombre, datos in personas.items()]

print(f'Edades: {Edades}')
print(f'Pesos: {Pesos}')
````



<b>Comparación entre for y comprehension</b>

La versión con for permite recorrer los datos paso a paso y entender más fácilmente cómo se van agregando los elementos a cada lista. Por otro lado, la comprensión de listas permite obtener el mismo resultado con menos líneas de código, haciendo que la extracción de las edades y los pesos sea más rápida de escribir.

En el caso del diccionario clasificacion, la comprensión de diccionarios también permite organizar las edades según sean mayores o menores de edad. Sin embargo, considero que la versión con for es más fácil de entender cuando se está aprendiendo Python, porque las condiciones se pueden identificar con mayor claridad.

Personalmente, preferiría utilizar comprehension cuando la operación es sencilla y quiero escribir un código más corto. En cambio, utilizaría for cuando hay varias condiciones o cuando necesito realizar diferentes operaciones, porque así el código resulta más fácil de leer, revisar y corregir.

<b>2. Resumen Vago: </b>

La **list comprehension** en Python permite crear listas de forma más corta y sencilla que con un ciclo `for`.

**Estructura**

```python
nueva_lista = [elemento for elemento in lista if condicion]
```

**En resumen**

La comprensión de listas se construye colocando primero lo que se quiere guardar, después el `for` que recorre los datos y, si es necesario, al final una condición `if` para filtrar los elementos.


<b>Resumen Especifico:</b> 

**Extracción de la lista Edades con List Comprehension**

La comprensión de listas (*list comprehension*) permite crear una lista de forma más corta y sencilla, sin necesidad de utilizar un ciclo `for` con `append()`.

```python
# Extrae la edad de cada persona y crea una lista
Edades = [datos[0] for nombre, datos in Personas.items()]

# Imprime la lista de edades
print(Edades)
```

**Explicación del código:**

* `Personas.items()`: permite recorrer los nombres y los datos de cada persona en el diccionario.
* `for nombre, datos in Personas.items()`: obtiene el nombre y la tupla de datos de cada persona.
* `datos[0]`: extrae la edad, ya que es el primer elemento de la tupla.
* Los corchetes `[]`: permiten crear la nueva lista con las edades obtenidas.

**Conclusión:** la comprensión de listas permite extraer las edades del diccionario utilizando menos líneas de código. Además, facilita la creación de listas de manera práctica y sencilla.

 ### ¿Qué es un REST web service?

 Un REST Web Service es un servicio web que permite que diferentes aplicaciones se comuniquen e intercambien información a través de internet, generalmente utilizando el protocolo HTTP. Utiliza direcciones URL para acceder a recursos y métodos como GET para consultar información y POST para enviar datos. Por ejemplo, una aplicación puede consultar el estado de un pedido mediante un servicio web.

 
 ### ¿Qué es una RESTful API?

 Una RESTful API es una interfaz que permite que dos aplicaciones se comuniquen siguiendo los principios de la arquitectura REST (Representational State Transfer). Permite consultar, crear, actualizar o eliminar información mediante métodos HTTP como GET, POST, PUT, PATCH y DELETE. Por ejemplo, una aplicación puede utilizar una API para consultar información de usuarios o registrar nuevos datos. Una RESTful API facilita el intercambio de información entre sistemas de manera organizada.
 
 ### ¿Qué es JSON?
 JSON (JavaScript Object Notation) es un formato de texto utilizado para almacenar e intercambiar información entre aplicaciones. Organiza los datos mediante pares de clave y valor, y también permite representar listas y estructuras de datos. Es fácil de leer tanto para las personas como para los programas.

```json
{
    "nombre": "Miguel",
    "edad": 24,
    "estudia": true
}
```

En este ejemplo, `nombre`, `edad` y `estudia` son las claves, mientras que `"Miguel"`, `24` y `true` son sus respectivos valores. JSON se utiliza frecuentemente para enviar y recibir información a través de las APIs.

````python

df1 = df.dropna(how='any')

df2 = df.dropna(how='all')


print("DataFrame original:", df.shape)
print("df1 (how='any'):", df1.shape)
print("df2 (how='all'):", df2.shape)
````

### Comparación de los DataFrames

Se crearon dos DataFrames aplicando diferentes métodos para eliminar los valores faltantes. El DataFrame `df1`, utilizando `dropna(how='any')`, elimina las filas que presentan al menos un dato nulo, mientras que `df2`, utilizando `dropna(how='all')`, elimina únicamente las filas en las que todos los valores son nulos.

Por esta razón, se espera que `df1` conserve menos filas que `df2`, ya que su condición de eliminación es más estricta. En cambio, `df2` conserva los registros que tienen información disponible, aunque algunas columnas contengan valores faltantes.

La comparación de ambos DataFrames permite observar cómo la estrategia utilizada para tratar los valores nulos afecta la cantidad de datos disponibles para el análisis.

````python
columnas = ['latitude', 'longitude']

df_base = df.copy()

for col in columnas:
    df_base[col] = pd.to_numeric(df_base[col], errors='coerce')

df_cero = df_base.copy()
df_cero[columnas] = df_cero[columnas].fillna(0)

df_ffill = df_base.copy()
df_ffill[columnas] = df_ffill[columnas].ffill()

df_promedio = df_base.copy()

for col in columnas:
    df_promedio[col] = df_promedio[col].fillna(df_promedio[col].mean())

print("Valores nulos originales:")
print(df_base[columnas].isna().sum())

print("\nValores nulos después de completar con 0:")
print(df_cero[columnas].isna().sum())

print("\nValores nulos después de ffill:")
print(df_ffill[columnas].isna().sum())

print("\nValores nulos después de completar con el promedio:")
print(df_promedio[columnas].isna().sum())
````

### Comparación de estrategias para completar datos faltantes

Se aplicaron tres métodos para completar los valores faltantes de las variables `latitude` y `longitude` en el DataFrame.

**1. Completar con el número 0:** este método reemplaza los valores nulos por cero. Es sencillo y permite completar los datos faltantes, pero no es recomendable para las coordenadas geográficas, porque puede ubicar incorrectamente a las cervecerías.

**2. Propagación hacia adelante (`ffill`):** este método reemplaza los valores faltantes con el último valor disponible de la misma columna. Permite conservar los registros, pero puede asignar coordenadas de otra cervecería y generar ubicaciones incorrectas. Si los primeros valores son nulos, estos pueden permanecer sin completar.

**3. Completar con el promedio:** este método reemplaza los valores faltantes por el promedio de cada columna. Permite completar los datos utilizando los valores disponibles, aunque puede asignar coordenadas que no corresponden a la ubicación real de ninguna cervecería.

las tres estrategias permiten tratar los datos faltantes, pero cada una tiene limitaciones. Para este caso, es importante revisar el efecto de cada método, ya que las coordenadas representan ubicaciones geográficas reales y una imputación inadecuada puede afectar los análisis y mapas posteriores.

### Análisis y conclusión

Rellenar las coordenadas geográficas con el promedio o con el número 0 puede no ser adecuado porque se pueden generar ubicaciones incorrectas. Por ejemplo, utilizar 0 en la latitud y longitud ubicaría el registro en una zona cercana al golfo de Guinea, mientras que el promedio puede representar un punto donde realmente no existe ninguna cervecería. Esto puede afectar los mapas y los análisis de ubicación. Por eso, antes de completar los datos, es importante revisar la información y elegir el método más apropiado.

</H3>

**Crear un dataframe grade_lookup**

````python
grade_lookup = pd.DataFrame({
    'grade_level': [1, 2, 3],
    'level': ['ELEMENTARY', 'MIDDLE/HIGH', 'MISC']
})

display(grade_lookup)
````

**Inner join (how='inner')**

````python
inner_join = pd.merge(
    df, grade_lookup,
    left_on='GRADE LEVEL',
    right_on='level',
    how='inner'
)

left_join = pd.merge(
    df, grade_lookup,
    left_on='GRADE LEVEL',
    right_on='level',
    how='left'
)

outer_join = pd.merge(
    df, grade_lookup,
    left_on='GRADE LEVEL',
    right_on='level',
    how='outer',
    indicator=True
)

print("INNER JOIN:", inner_join.shape)
display(inner_join.head())

print("LEFT JOIN:", left_join.shape)
display(left_join.head())

print("OUTER JOIN:", outer_join.shape)
display(outer_join.head())
````

**INNER JOIN (`how='inner'`)**

Esta unión conserva únicamente los registros que tienen coincidencias entre la columna `GRADE LEVEL` del DataFrame original y la columna `level` del DataFrame `grade_lookup`. En este caso, coinciden los niveles `ELEMENTARY` y `MIDDLE/HIGH`. Los registros con `DISTRICT TOTAL` no se conservan porque ese valor no existe en la tabla de equivalencias.

**LEFT JOIN (`how='left'`)**

Esta unión conserva todos los registros del DataFrame original, incluso aquellos que no tienen coincidencia en la tabla de equivalencias. Las filas correspondientes a `DISTRICT TOTAL` se mantienen, pero la columna `grade_level` queda con un valor nulo porque no existe una equivalencia para ese nivel.

**OUTER JOIN (`how='outer'`)**

Esta unión conserva todos los registros de ambos DataFrames, tengan o no coincidencias. Por eso, incluye los registros de `DISTRICT TOTAL` del DataFrame original y el nivel `MISC` de la tabla de equivalencias, aunque no coincidan entre sí. La columna `_merge` permite identificar cuáles registros coinciden en ambos DataFrames y cuáles aparecen únicamente en uno de ellos.

**Conclusión**

Los tres tipos de unión permiten combinar información de diferentes maneras. `inner` conserva únicamente las coincidencias, `left` mantiene todos los registros del DataFrame principal y `outer` reúne los registros de ambos conjuntos. Para evitar valores sin correspondencia, es importante revisar las categorías y verificar que los valores utilizados como claves tengan equivalencias correctas.

**Realiar un groupby() para suma de obesos por grado**

````python

suma_obesos = df.groupby('GRADE LEVEL')['NO. OBESE'].sum()

display(suma_obesos)
````

````python

resumen_obesos = df.groupby('GRADE LEVEL')['NO. OBESE'].agg(
    ['sum', 'mean', 'std']
)

display(resumen_obesos)
````

**Crear una pivot_table()**

````python
tabla_obesos = pd.pivot_table(
    df,
    index='GRADE LEVEL',
    columns='AREA TYPE',
    values='NO. OBESE',
    aggfunc='sum',
    fill_value=0
)

display(tabla_obesos)
````

**Comparación entre `groupby()` y `pivot_table()`**

`groupby()` y `pivot_table()` permiten agrupar y resumir datos, pero se diferencian en la forma en que presentan los resultados.

- **`groupby()`** es más conveniente cuando se necesita agrupar los datos por una o varias variables y calcular operaciones como la suma, el promedio o la desviación estándar. Por ejemplo, calcular el total de estudiantes obesos por nivel educativo.

- **`pivot_table()`** es más conveniente cuando se desea comparar los resultados entre dos o más categorías en una tabla cruzada, organizando una variable en las filas y otra en las columnas. Por ejemplo, comparar el total de estudiantes obesos según el nivel educativo y el tipo de área.
